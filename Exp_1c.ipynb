```python
import pandas as pd

structured = pd.DataFrame({
    'Name': ['Arun', 'Priya', 'Ravi'],
    'Age': [20, 21, 22]
})

unstructured = [
    'This is a text document.',
    'This is an image file.',
    'This is an audio recording.'
]

semi_structured = {
    'student1': {'Name': 'Arun', 'Age': 20},
    'student2': {'Name': 'Priya', 'Age': 21}
}

print("Structured Data:")
print(structured)

print("\nUnstructured Data:")
for item in unstructured:
    print(item)

print("\nSemi-Structured Data:")
print(pd.json_normalize(semi_structured).T)
```
