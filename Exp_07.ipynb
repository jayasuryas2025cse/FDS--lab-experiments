```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

population = np.random.normal(50, 10, 100000)
sample_sizes = [30, 50, 100]
num_samples = 1000

plt.figure(figsize=(10, 8))

for i, size in enumerate(sample_sizes):
    sample_means = []

    for _ in range(num_samples):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means.append(np.mean(sample))

    plt.subplot(3, 1, i + 1)
    plt.hist(sample_means, bins=30, alpha=0.7)
    plt.axvline(np.mean(population), color='red', linestyle='dashed')
    plt.title(f'Sampling Distribution (Sample Size {size})')
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
