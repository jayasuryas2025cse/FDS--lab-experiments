```python
import pandas as pd
import numpy as np

data = {
    'CustomerID': [1, 2, 3, 4, 5],
    'Age': [22, 32, np.nan, 25, 40],
    'Rating': [4, 5, 6, -1, 3],
    'Hotel': ['Ibis', 'LemonTree', 'RedFox', 'Ibis', None],
    'Bill': [1300, 2000, 1322, -500, 1800],
    'NoOfPax': [2, 3, 2, 0, 4]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Hotel'] = df['Hotel'].fillna(df['Hotel'].mode()[0])
df.loc[~df['Rating'].between(1, 5), 'Rating'] = np.nan
df.loc[df['Bill'] < 0, 'Bill'] = np.nan
df.loc[df['NoOfPax'] <= 0, 'NoOfPax'] = np.nan

df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Bill'] = df['Bill'].fillna(df['Bill'].median())
df['NoOfPax'] = df['NoOfPax'].fillna(df['NoOfPax'].median())

print("\nCleaned Data:")
print(df)
print("\nMissing Values:")
print(df.isnull().sum())
```
