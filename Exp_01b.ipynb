```python
import pandas as pd
import matplotlib.pyplot as plt

roles = ['Data Analyst', 'Data Engineer', 'Data Scientist', 'ML Engineer', 'BI Analyst']
counts = [30, 25, 35, 20, 15]

df = pd.DataFrame({
    'Role': roles,
    'Count': counts
})

print(df)

plt.bar(df['Role'], df['Count'])
plt.title('Distribution of Data Science Roles')
plt.xlabel('Job Role')
plt.ylabel('Number of Postings')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
