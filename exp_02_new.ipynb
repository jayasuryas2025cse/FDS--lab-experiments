```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Product': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Sales': [100, 200, 150, 300, 250, 350],
    'Quantity': [2, 4, 3, 6, 5, 7],
    'Region': ['East', 'West', 'East', 'North', 'West', 'South']
})

print("Dataset:")
print(df)

print("\nSummary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSales by Product:")
sales = df.groupby('Product')['Sales'].sum()
print(sales)

sales.plot(kind='bar', title='Sales by Product')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.show()

df.plot(x='Quantity', y='Sales', kind='scatter', title='Sales vs Quantity')
plt.show()
```
