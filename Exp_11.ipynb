```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Date': pd.date_range(start='2023-01-01', periods=16, freq='D'),
    'Product': ['Product A', 'Product B', 'Product A', 'Product C',
                'Product B', 'Product A', 'Product C', 'Product B',
                'Product A', 'Product C', 'Product B', 'Product A',
                'Product C', 'Product B', 'Product C', 'Product A'],
    'Sales': [200, 150, 220, 300, 180, 250, 320, 170, 280, 310, 190, 260, 340, 160, 330, 220],
    'Quantity': [4, 3, 5, 6, 4, 7, 8, 3, 6, 7, 4, 5, 8, 3, 7, 6],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
               'North', 'East', 'South', 'North', 'East', 'West', 'East', 'North']
}

df = pd.DataFrame(data)

print(df.head())
print(df.isnull().sum())
print(df.describe())

print(df.groupby('Product')[['Sales', 'Quantity']].sum())

pivot_table = df.pivot_table(values='Sales', index='Region', columns='Product', aggfunc='sum', fill_value=0)
print(pivot_table)

correlation_matrix = df[['Sales', 'Quantity']].corr()
print(correlation_matrix)

sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Matrix')
plt.show()

df.groupby('Date')['Sales'].sum().plot()
plt.title('Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.show()
```
