```python
from collections import Counter
import matplotlib.pyplot as plt

text = """
Data science uses data to solve problems.
Data analysis helps understand data.
Python is useful for data science.
"""

words = text.lower().split()
frequency = Counter(words)

print(frequency)

common_words = frequency.most_common()
labels = [item[0] for item in common_words]
values = [item[1] for item in common_words]

plt.bar(labels, values)
plt.title('Word Frequency')
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
